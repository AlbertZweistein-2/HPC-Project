```python
import math
```

```python
def AllGather(r, p):
    q = math.ceil(math.log2(p))
    print(f"Set B[0] = {r} in process {r}")
    for k in range(q):
        
        t, f = (r - 2**k + p) % p, (r + 2**k) % p
        print(f"Send B[0..{2**(k+1)-2**(k)-1}] to process {t} from process {r}")
        print(f"Receive B[{2**(k)}..{2**(k+1)-1}] from process {f} to process {r}")

    for i in range(1, p):
        print(f"Set R[{(r+i)%p}] to B[{i}]")
```

```python
p = 5
for r in range(0, p):
    AllGather(r, p)
    print(f"Process {r} has completed AllGather operation.\n")
```

```python
p = 15
for r in range(0, p):
    AllGather(r, p)
    print(f"Process {r} has completed AllGather operation.\n")
```

```python
def f(origin, r, p):
    q = math.ceil(math.log2(p))
    k = q-1
    s_k = 2**k - ((2**q) - p)
    return ((origin - r + p) % p) < s_k
```

```python
p = 17
r = 0
for i in range(0, p):
    print(f"f({i}, {r}, {p}) = {f(i, r, p)}")
```
